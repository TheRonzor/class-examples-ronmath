## Complex Numbers

These have real and imaginary components, both defined as `Rational`s. It can also accept text input!

```python
w = Complex.from_string('2.1 + 3.14i')
w
```

```python
# Complex conjugate
w.conjugate()
```

```python
# Unfortunately, I didn't implement multiplication by regular Python numbers:
5*w
```

```python
# But it's easy enough:
multiple = 5
result = Complex.zero()

for _ in range(multiple):
    result += w
result
```

```python
result.reduce()
result
```

## Matrices

```python
X = Matrix([
    [x, y],
    [z, Rational.one()]
])
X
```

```python
X = X @ X
X
```

```python
X.reduce()
X
```

```python
X.T
```

```python
Z = ComplexMatrix([
    [w+w, w-w],
    [w*w, w/w]
])
Z
```

```python
Z.reduce()
Z
```

```python
# Conjugate transpose
Z.CT
```
